# 1.Bibliotecas

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from scipy import stats
```

#2. Preparação dos dados

```python
df = pd.read_csv('../data/processed/churn_tratado.csv')
```

#3. Analise estatistica das variaveis numericas

##3.1 **tenure** x **Churn**

```python
tenure_churn = df[df['Churn']=='Yes']['tenure']
tenure_no_churn = df[df['Churn']=='No']['tenure']
```

```python
print(tenure_churn.describe())
print(tenure_no_churn.describe())
```

```python
len(tenure_churn), len(tenure_no_churn)
```

```python
stats.mannwhitneyu(
    tenure_churn,
    tenure_no_churn,
    alternative='two-sided'
)
```

### Resultado

O teste de Mann–Whitney tem `p-value < 0,05`, mostrando que há diferença entre as distribuições de **tenure** dos grupos `Churn = Yes` e `Churn = No`.

Esse resultado confirma estatisticamente a diferença que vimos durante a análise exploratória, onde os clientes que realizaram churn apresentaram valores menor **tenure**.

Ainda não podemos dizer que tenure causa churn.

# 4. Analise estatistica das variaveis categoricas

## 4.1 **Contract** × **Churn**

```python
tabela_cont_C = pd.crosstab(
    df['Contract'],
    df['Churn'],
)
tabela_cont_C
```

```python
chi2, p_value, dof, expected = stats.chi2_contingency(tabela_cont_C)
chi2, p_value,dof, expected
```

### Resultado

Qui-quadrado tem `p-value < 0,05`, mostrando que existe associação entre **Contract** e **Churn**.

Os valores vistos também tem diferenças consideráveis em relação aos valores esperados caso as variáveis fossem independentes. Oque pode ser observado ao olhar para o valor `Expected`

Esse resultado mostra que tem diferença estatistica, assim como foi observado na análise exploratória.
